1. Construya un analizador que recorra varios archivos `.txt`, extraiga información mediante expresiones regulares y genere un reporte consolidado. Considere los archivos disponibles en `Data`, entre ellos `texto_1.txt`, `texto_2.txt`, `texto_3.txt`, etc...

1. Define una función llamada `analizar_archivo(ruta_archivo)` que reciba la ruta de un archivo `.txt` y devuelva un diccionario con:
   - nombre del archivo;
   - número de líneas;
   - número de palabras;
   - cantidad de números encontrados;
   - cantidad de números decimales;
   - cantidad de palabras que comienzan con vocal.
   
2. Utiliza `with open(..., encoding='utf-8')` para leer cada archivo. Recorre únicamente los archivos que tengan extensión `.txt`. Ordénalos por nombre para que el análisis sea reproducible.

3. Define y aplica los siguientes patrones con el módulo `re`:
   - **Número entero o decimal:** `r'(?<![\w.])-?\d+(?:[.,]\d+)?(?![\w.])'`
   - **Número decimal:** `r'(?<![\w.])-?\d+[.,]\d+(?![\w.])'`
   - **Palabra que comienza con vocal:** `r'\b[aeiouáéíóúü][a-záéíóúüñ]*\b'`, usando `re.IGNORECASE`.
4. Convierte los resultados en un `DataFrame` y muestra una fila por archivo.


5. Conteste las siguientes Preguntas 
    1. ¿Qué archivo contiene más líneas y cuál contiene más palabras?
    2. ¿Cuántos números enteros o decimales se encontraron en total?
    3. ¿Qué archivo contiene más números decimales?
    4. ¿Qué archivo tiene la mayor proporción de palabras que contiene vocales?


```python
import os
import re
import pandas as pd

dir_base = os.getcwd()
ruta = os.path.join(dir_base, 'Data')
print(ruta)
```

```python
patron_numero = r'(?<![\w.])-?\d+(?:[.,]\d+)?(?![\w.])'
patron_decimal = r'(?<![\w.])-?\d+[.,]\d+(?![\w.])'
patron_vocal = r'\b[aeiouáéíóú][a-záéíóúñ]*\b'


def analizar_archivo(ruta_archivo):
    with open(ruta_archivo, encoding='utf-8') as f:
        contenido = f.read()

    lineas = contenido.splitlines()
    palabras = contenido.split()

    numeros = re.findall(patron_numero, contenido)
    decimales = re.findall(patron_decimal, contenido)
    palabras_vocal = re.findall(patron_vocal, contenido, flags=re.IGNORECASE)

    return {
        'archivo': os.path.basename(ruta_archivo),
        'num_lineas': len(lineas),
        'num_palabras': len(palabras),
        'num_numeros': len(numeros),
        'num_decimales': len(decimales),
        'num_palabras_vocal': len(palabras_vocal)
    }
```

```python
archivos_txt = sorted([
    archivo for archivo in os.listdir(ruta)
    if archivo.endswith('.txt')
])
print(archivos_txt)
```

```python
resultados = [analizar_archivo(os.path.join(ruta, archivo)) for archivo in archivos_txt]
df_reporte = pd.DataFrame(resultados)
df_reporte
```

```python
archivo_mas_lineas = df_reporte.loc[df_reporte['num_lineas'].idxmax(), 'archivo']
archivo_mas_palabras = df_reporte.loc[df_reporte['num_palabras'].idxmax(), 'archivo']
print(f"1. Archivo con más líneas: {archivo_mas_lineas}")
print(f"   Archivo con más palabras: {archivo_mas_palabras}")

total_numeros = df_reporte['num_numeros'].sum()
print(f"2. Total de números enteros o decimales encontrados: {total_numeros}")

archivo_mas_decimales = df_reporte.loc[df_reporte['num_decimales'].idxmax(), 'archivo']
print(f"3. Archivo con más números decimales: {archivo_mas_decimales}")

df_reporte['proporcion_vocal'] = df_reporte['num_palabras_vocal'] / df_reporte['num_palabras']
archivo_mayor_proporcion = df_reporte.loc[df_reporte['proporcion_vocal'].idxmax(), 'archivo']
print(f"4. Archivo con mayor proporción de palabras que comienzan con vocal: {archivo_mayor_proporcion}")
```
